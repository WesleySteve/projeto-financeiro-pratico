# Analise arquivos raw vendas.csv

```python
import os
import pandas as pd
from pandasql import sqldf

pysqldf = lambda q: sqldf(q, globals())
```

```python
BASE_DIR = os.path.dirname(os.path.dirname(os.path.dirname(os.path.abspath('.'))))
DATA_DIR = os.path.join(BASE_DIR, 'data')
RAW_DIR =  os.path.join(DATA_DIR, 'raw')
```

```python
df = pd.read_csv(os.path.join(RAW_DIR, 'vendas.csv'), sep=';')
df.shape
```

```python
q = '''SELECT id_venda
            , id_cliente
            , id_produto
            , quantidade
            , valor_unitario
            , desconto
            , forma_pagamento
            , status_venda
            , valor_total
        FROM df
        WHERE id_cliente isnull
            or id_produto isnull
            or quantidade isnull
            or valor_unitario isnull
            or desconto isnull
            or forma_pagamento isnull
            or status_venda isnull
            or status_venda isnull
            or valor_total 

'''
pysqldf(q)
```
