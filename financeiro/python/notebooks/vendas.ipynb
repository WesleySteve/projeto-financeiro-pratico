## Tratamento de dados

```python
df_original = df.copy()
```

```python
df = df.fillna('NAO INFORMADO')
```

```python
# Remove espaços antes/depois dos nomes
df.columns = df.columns.str.strip()
```

```python
df.info()
```

```python
campos_texto = [
    "id_venda",
    "id_cliente",
    "id_produto",
    "forma_pagamento",
    "forma_pagamento",
    "status_venda"
]
```

```python
for campo in campos_texto:

    df[campo] = (
        df[campo]
        .astype("string")
        .str.strip()
    )
```

```python
campos_numericos = [
    "quantidade",
    "valor_unitario",
    "desconto",
    "valor_total"
]
```

```python
for campo in campos_numericos:
    df[campo] = (
        df[campo].replace('NAO INFORMADO', 0)
        .astype("float")
    )
```

```python
# campos datas

df['data_venda'] = df['data_venda'].replace('NAO INFORMADO', '1900-01-01', regex=True)
df['data_venda'] = df['data_venda'].replace('/', '-',  regex=True)

```

```python
df['data_venda'] = pd.to_datetime(df['data_venda'], format='mixed')
```
